```python
import pandas as pd
```

```python
df=pd.read_csv("Telco-Customer-Churn.csv")
```

```python
df=df[df['TotalCharges']!=' ']
df['TotalCharges']=df['TotalCharges'].astype(float)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
columns=['OnlineSecurity','OnlineBackup','DeviceProtection','TechSupport','StreamingTV','StreamingMovies','MultipleLines']
for col in columns:
    df[col]=df[col].replace({'No internet service':'No'})
    df[col]=df[col].replace({'No phone service':'No'})
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder
le=LabelEncoder()
ohe=OneHotEncoder()
```

```python
columns=['OnlineSecurity','OnlineBackup','DeviceProtection','TechSupport','StreamingTV','StreamingMovies','MultipleLines','gender','Partner','Dependents','PhoneService','PaperlessBilling']
for col in columns:
  df[col]=le.fit_transform(df[col])
```

```python
df
```

```python
encoded_df=pd.get_dummies(df, columns=["InternetService","Contract","PaymentMethod"], dtype=int)
```

```python
encoded_df=encoded_df.drop(["customerID","InternetService_DSL","Contract_Month-to-month","PaymentMethod_Bank transfer (automatic)"], axis=1)
```

```python
encoded_df['Churn']=le.fit_transform(encoded_df['Churn'])
```

```python
encoded_df
```

```python
encoded_df.info()
```

```python
input=encoded_df.drop("Churn", axis=1)
output=encoded_df["Churn"]
```

```python
input
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
X_train, X_test, y_train, y_test=train_test_split(input, output, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```

```python
model=LogisticRegression()
model.fit(X_train, y_train)
```

```python
model.score(X_train,y_train)
```

```python
predict=model.predict(X_test)
```

```python
predict
```

```python
from sklearn.metrics import accuracy_score
print(accuracy_score(predict,y_test))
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test, predict)
cm
```

```python
import seaborn as sns
```

```python
sns.heatmap(cm, annot=True)
```

```python
from sklearn.metrics import precision_score
print(precision_score(y_test, predict))
```

```python
from sklearn.metrics import recall_score
print(recall_score(y_test, predict))
```

```python
from sklearn.metrics import f1_score
print(f1_score(y_test, predict))
```

```python
from sklearn.ensemble import RandomForestClassifier
randomForest=RandomForestClassifier(n_estimators=250)
randomForest.fit(X_train, y_train)
```

```python
predhict2=randomForest.predict(X_test)
```

```python
accuracy_score(y_test, preict2)
```

```python
from sklearn.model_selection import GridSearchCV
param={'n_estimators':[100,200,300,250,400]}
grid=GridSearchCV(randomForest, param, cv=5)
grid.fit(X_train, y_train)
```
